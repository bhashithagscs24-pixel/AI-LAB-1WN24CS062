```python

import random
import math

def calculate_cost(board):
    h = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                h += 1

    return h


def print_board(board):
    n = len(board)

    for i in range(n):
        for j in range(n):
            if board[j] == i:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def simulated_annealing(board):
    temperature = 100
    cooling_rate = 0.995
    max_iterations = 100000

    current = board[:]
    current_cost = calculate_cost(current)

    best = current[:]
    best_cost = current_cost

    for step in range(max_iterations):

        if current_cost == 0:
            print("\nSolution Found!")
            print("Iterations:", step)
            return current

        neighbor = current[:]

        col = random.randint(0, 7)
        row = random.randint(0, 7)

        while row == neighbor[col]:
            row = random.randint(0, 7)

        neighbor[col] = row
        neighbor_cost = calculate_cost(neighbor)

        delta = neighbor_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / temperature):
            current = neighbor
            current_cost = neighbor_cost

        if current_cost < best_cost:
            best = current[:]
            best_cost = current_cost

        temperature *= cooling_rate

        if temperature < 0.0001:
            break

    if best_cost == 0:
        print("\nSolution Found!")
        print("Best Cost:", best_cost)
        return best
    else:
        print("\nSolution Not Found!")
        print("Best Cost:", best_cost)
        return None


print("Enter the initial positions of 8 queens (1 to 8):")

board = []

for i in range(8):
    row = int(input("Enter row for column " + str(i + 1) + ": "))
    board.append(row - 1)

if any(row < 0 or row > 7 for row in board):
    print("Invalid input! Enter values between 1 and 8.")
else:
    print("\nInitial Board:")
    print_board(board)

    print("\nInitial Cost:", calculate_cost(board))

    solution = simulated_annealing(board)

    if solution:
        print("\nFinal Configuration:", [x + 1 for x in solution])
        print("\nFinal Board:")
        print_board(solution)
        print("Final Cost:", calculate_cost(solution))
337
```
