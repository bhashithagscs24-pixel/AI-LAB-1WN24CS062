```python

facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}
print("BHASHITHA G S ")
print("1WN24CS062")

new = True

while new:
    new = False

    for fact in list(facts):
        if fact[0] == "Missile":
            weapon = ("Weapon", fact[1])
            if weapon not in facts:
                facts.add(weapon)
                new = True

    if ("Enemy", "A", "America") in facts:
        if ("Hostile", "A") not in facts:
            facts.add(("Hostile", "A"))
            new = True

    for fact in list(facts):
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in facts:
                sells = ("Sells", "Robert", x, "A")
                if sells not in facts:
                    facts.add(sells)
                    new = True

    for fact in list(facts):
        if fact[0] == "American":
            p = fact[1]

            for weapon in list(facts):
                if weapon[0] == "Weapon":
                    q = weapon[1]

                    for sale in list(facts):
                        if sale[0] == "Sells":
                            if sale[1] == p and sale[2] == q:
                                r = sale[3]

                                if ("Hostile", r) in facts:
                                    criminal = ("Criminal", p)

                                    if criminal not in facts:
                                        facts.add(criminal)
                                        new = True

print("Derived Facts:")
for fact in sorted(facts):
    print(fact)

if ("Criminal", "Robert") in facts:
    print("\nRobert is a Criminal")
else:
    print("\nRobert is not proved to be a Criminal")
```
