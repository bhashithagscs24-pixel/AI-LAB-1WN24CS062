```python

facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}
print("BHASHITHA G S ")
print("1WN24CS062")

new = True

while new:
    new = False

    for fact in list(facts):
        if fact[0] == "Missile":
            weapon = ("Weapon", fact[1])
            if weapon not in facts:
                facts.add(weapon)
                new = True

    if ("Enemy", "A", "America") in facts:
        if ("Hostile", "A") not in facts:
            facts.add(("Hostile", "A"))
            new = True

    for fact in list(facts):
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in facts:
                sells = ("Sells", "Robert", x, "A")
                if sells not in facts:
                    facts.add(sells)
                    new = True

    for fact in list(facts):
        if fact[0] == "American":
            p = fact[1]

            for weapon in list(facts):
                if weapon[0] == "Weapon":
                    q = weapon[1]

                    for sale in list(facts):
                        if sale[0] == "Sells":
                            if sale[1] == p and sale[2] == q:
                                r = sale[3]

                                if ("Hostile", r) in facts:
                                    criminal = ("Criminal", p)

                                    if criminal not in facts:
                                        facts.add(criminal)
                                        new = True

print("Derived Facts:")
for fact in sorted(facts):
    print(fact)

if ("Criminal", "Robert") in facts:
    print("\nRobert is a Criminal")
else:
    print("\nRobert is not proved to be a Criminal")
```

```python

facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

print("Initial Facts:")
for f in sorted(facts):
    print(f[0] + "(" + ", ".join(f[1:]) + ")")

iteration = 1
new = True

while new:
    new = False
    new_facts = set()

    for f in facts:
        if f[0] == "Missile":
            new_facts.add(("Weapon", f[1]))

    if ("Enemy", "A", "America") in facts:
        new_facts.add(("Hostile", "A"))

    for f in facts:
        if f[0] == "Missile":
            x = f[1]
            if ("Owns", "A", x) in facts:
                new_facts.add(("Sells", "Robert", x, "A"))

    for f in facts:
        if f[0] == "American":
            p = f[1]

            for w in facts:
                if w[0] == "Weapon":
                    q = w[1]

                    for s in facts:
                        if s[0] == "Sells" and s[1] == p and s[2] == q:
                            r = s[3]

                            if ("Hostile", r) in facts:
                                new_facts.add(("Criminal", p))

    new_facts = new_facts - facts

    if new_facts:
        print("\nIteration", iteration, ":")
        for f in sorted(new_facts):
            print(f[0] + "(" + ", ".join(f[1:]) + ")")

        facts.update(new_facts)
        new = True
        iteration += 1

print("\nFinal Facts:")
for f in sorted(facts):
    print(f[0] + "(" + ", ".join(f[1:]) + ")")

if ("Criminal", "Robert") in facts:
    print("\nConclusion: Robert is a Criminal.")
```
