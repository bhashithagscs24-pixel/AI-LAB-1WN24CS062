```python

from itertools import product

def implies(a, b):
    return not a or b

def truth_table():
    print("P Q R | Q->P | P->~Q | Q OR R | KB")
    print("--------------------------------------")

    models = []

    for P, Q, R in product([True, False], repeat=3):

        s1 = implies(Q, P)
        s2 = implies(P, not Q)
        s3 = Q or R

        KB = s1 and s2 and s3

        print(int(P), int(Q), int(R), "|",
              int(s1), "   ",
              int(s2), "   ",
              int(s3), "    ",
              int(KB))

        if KB:
            models.append((P, Q, R))

    print("\nModels where KB is True:")

    for model in models:
        print(model)

    print("\nKB entails R:", all(R for P, Q, R in models))

    print("KB entails R -> P:",
          all(implies(R, P) for P, Q, R in models))

    print("KB entails Q -> R:",
          all(implies(Q, R) for P, Q, R in models))


truth_table()
```
